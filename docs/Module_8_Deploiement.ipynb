## 1. Pourquoi déployer ?

> **Le fossé « ça marche sur ma machine ».** Un piège classique : l'application fonctionne chez toi mais plante en ligne, parce que le serveur n'a pas les mêmes bibliothèques, ni les mêmes secrets, ni la même configuration. Tout ce module vise à **combler ce fossé** : lister les dépendances, externaliser les secrets, adapter la configuration.

## 2. Git — le contrôle de version

**Git** est un outil qui enregistre l'**historique** de ton code : chaque version, qui a changé quoi, et permet de revenir en arrière. C'est indispensable, et c'est aussi le moyen d'envoyer ton code vers un hébergeur.

Vocabulaire de base :

| Terme | Signification |
|---|---|
| **dépôt** (*repository*, « repo ») | Le dossier suivi par Git (ton projet + son historique) |
| **commit** | Un « instantané » enregistré de ton code à un moment donné |
| **add** (*stage*) | Sélectionner les fichiers à inclure dans le prochain commit |

Les commandes essentielles, **à taper dans ton terminal** (dans le dossier de ton projet) :

```bash
git init                       # initialise Git dans le dossier (une seule fois)
git status                     # montre les fichiers modifiés / à ajouter
git add .                      # ajoute TOUS les fichiers modifiés au prochain commit
git add app.py                 # ou seulement un fichier précis
git commit -m "Mon premier commit"   # enregistre un instantané avec un message
git log                        # affiche l'historique des commits
```

> **Le cycle de travail Git :** tu modifies des fichiers → `git add` (tu sélectionnes) → `git commit` (tu enregistres). Répète à chaque étape importante de ton projet. Un bon message de commit décrit *ce que tu as changé* (« Ajoute la page de connexion »).

*(Ces commandes ne s'exécutent pas dans le notebook — recopie-les dans PowerShell ou le terminal VS Code.)*

## 3. GitHub — publier son code en ligne

**Git** est l'outil local ; **GitHub** est un service web qui **héberge tes dépôts** dans le cloud. Publier sur GitHub permet de sauvegarder ton code, de collaborer, et surtout de **connecter ton projet à un hébergeur** (qui ira y chercher le code).

Ne confonds pas :

| Git | GitHub |
|---|---|
| Un **outil** installé sur ta machine | Un **site web** (github.com) |
| Gère l'historique en local | Stocke les dépôts en ligne |
| Fonctionne sans internet | Nécessite un compte |

Étapes pour publier (une fois ton compte GitHub créé) :

1. Sur github.com, clique **New repository**, donne-lui un nom (ex. `mon-blog`), **sans** cocher « Add a README » (ton dossier existe déjà).
2. GitHub t'affiche des commandes. Dans ton terminal, relie ton dépôt local au dépôt distant et envoie ton code :

```bash
git remote add origin https://github.com/ton-pseudo/mon-blog.git
git branch -M main                 # nomme la branche principale "main"
git push -u origin main            # envoie (push) ton code vers GitHub
```

Ensuite, à chaque nouvelle modification, le cycle devient :

```bash
git add .
git commit -m "Description du changement"
git push                           # envoie les nouveaux commits sur GitHub
```

| Terme | Signification |
|---|---|
| **remote** (`origin`) | Le dépôt distant (sur GitHub) lié à ton dépôt local |
| **push** | Envoyer tes commits locaux vers le dépôt distant |
| **clone** | Copier un dépôt distant sur une nouvelle machine (`git clone <url>`) |

> **Astuce :** GitHub ne demande plus de mot de passe pour `push`, mais un **token** (ou une connexion via l'appli GitHub / GitHub CLI). Si `git push` échoue en demandant un mot de passe, crée un *Personal Access Token* dans les réglages GitHub, ou installe **GitHub CLI** (`gh auth login`).

## Variables d'environnement, fichier `.env` et `python-dotenv`

Les **variables d'environnement** permettent de stocker les informations de configuration d'une application (clés secrètes, mots de passe, paramètres de base de données, clés API, etc.) en dehors du code source. Cette approche améliore la **sécurité**, facilite la **configuration** et permet d'utiliser le même code dans différents environnements (développement, test et production).

Le **fichier `.env`** est un simple fichier texte qui contient ces variables sous la forme `NOM_VARIABLE=valeur`. Il est principalement utilisé pendant le développement et ne doit pas être partagé publiquement. Il est recommandé de l'ajouter au fichier `.gitignore` afin d'éviter de publier des informations sensibles sur GitHub.

La bibliothèque **`python-dotenv`** permet de charger automatiquement les variables définies dans le fichier `.env` dans l'environnement de l'application Python grâce à la fonction `load_dotenv()`. Les valeurs peuvent ensuite être récupérées avec `os.getenv("NOM_VARIABLE")`.

**Exemple de fichier `.env` :**

```text
SECRET_KEY=mon_secret
DEBUG=True
DB_HOST=localhost
DB_USER=postgres
DB_PASSWORD=123456

````python
from dotenv import load_dotenv
import os

load_dotenv()

secret_key = os.getenv("SECRET_KEY")
debug = os.getenv("DEBUG")
````

> **L'intérêt concret :** ton code publié sur GitHub contient `os.environ.get("SECRET_KEY")` — **pas** la clé elle-même. La vraie clé vit uniquement dans les variables d'environnement du serveur (et dans ton `.env` local, non publié). Le secret n'apparaît **jamais** dans le code.

````python
import os
from dotenv import load_dotenv

load_dotenv()   # lit le fichier .env et remplit os.environ

print("SECRET_KEY   :", os.environ.get("SECRET_KEY"))
print("DATABASE_URL :", os.environ.get("DATABASE_URL"))
print("DEBUG        :", os.environ.get("DEBUG"))
````

> **Bonne pratique : le fichier `.env.example`.** Comme `.env` n'est pas publié (secrets), on publie à sa place un fichier `.env.example` qui liste les **noms** des variables **sans les valeurs**. Ainsi, quelqu'un qui récupère ton projet sait quelles variables créer.

```
SECRET_KEY=
DATABASE_URL=
DEBUG=
```

## 8. Adapter son application pour la production

Ton `app.py` fonctionne en local, mais trois choses doivent changer pour la production. On applique tout ce qu'on vient de voir :

| En local (dev) | En production | Pourquoi |
|---|---|---|
| `app.secret_key = "en dur"` | lue depuis `os.environ` | ne jamais publier le secret |
| `app.run(debug=True)` | `debug=False` | le mode debug expose des infos sensibles en cas d'erreur |
| `app.run(port=5000)` | port lu depuis `os.environ["PORT"]` | l'hébergeur impose son propre port |

Voici la version « prête pour la production » de ton `app.py`. La cellule l'écrit sur le disque.

````python
%%writefile app.py
import os
from flask import Flask, jsonify
from dotenv import load_dotenv

load_dotenv()   # charge le .env en local (ignoré en prod, où les vars viennent du système)

app = Flask(__name__)

# 1. Secret lu depuis l'environnement (jamais en dur)
app.secret_key = os.environ.get("SECRET_KEY", "cle-de-secours-dev")

# 2. Base de données : configurable selon l'environnement
app.config["SQLALCHEMY_DATABASE_URI"] = os.environ.get("DATABASE_URL", "sqlite:///blog.db")

@app.route("/")
def accueil():
    return jsonify({"message": "Mon blog est en ligne !"})

if __name__ == "__main__":
    # 3. Port et debug lus depuis l'environnement
    port = int(os.environ.get("PORT", 5000))
    debug = os.environ.get("DEBUG", "False") == "True"
    app.run(host="0.0.0.0", port=port, debug=debug)
````

Points clés de cette version :
- `host="0.0.0.0"` rend l'app accessible depuis l'extérieur (nécessaire en production ; en local, `127.0.0.1` suffisait).
- `int(os.environ.get("PORT", 5000))` : le port arrive comme **texte** depuis l'environnement, on le convertit en nombre.
- `debug` ne vaut `True` que si la variable `DEBUG` vaut exactement `"True"` — donc `False` par défaut en production.

Vérifions que cette app démarre correctement et lit bien sa configuration (test **exécutable**, sans lancer de serveur) :

````python
import importlib
import app as app_module
importlib.reload(app_module)

client = app_module.app.test_client()
r = client.get("/")
print("Réponse de l'app :", r.status_code, r.get_json())
print("Secret chargé (non vide) :", bool(app_module.app.secret_key))
print("URI base de données :", app_module.app.config["SQLALCHEMY_DATABASE_URI"])
````

## 9. Le serveur de production et le `Procfile`

Le serveur intégré de Flask (`app.run()`) affiche d'ailleurs un avertissement : *« WARNING: This is a development server. Do not use it in a production deployment. »* Il n'est pas conçu pour encaisser le trafic réel.

En production, on utilise un **vrai serveur WSGI**. Le plus courant est **gunicorn**. On indique à l'hébergeur comment lancer l'app via un fichier **`Procfile`** (sans extension) :

````python
%%writefile Procfile
web: gunicorn app:app
````

````python
print(open("Procfile", encoding="utf-8").read())
````

Décryptage de `web: gunicorn app:app` :
- `web:` = ce processus répond aux requêtes web ;
- `gunicorn` = le serveur de production ;
- `app:app` = dans le fichier `app.py`, utilise l'objet nommé `app` (`fichier:variable`).

> **⚠️ Note spéciale Windows (ton cas).** `gunicorn` ne fonctionne **que sous Linux/Mac**. Ce n'est pas un problème : les hébergeurs (Render, Railway…) tournent **sous Linux**, donc gunicorn s'y installera parfaitement via ton `requirements.txt`. Mais si tu veux tester un serveur de production **en local sous Windows**, utilise plutôt **waitress** (`pip install waitress`, puis `waitress-serve --port=5000 app:app`). Pour le déploiement, garde `gunicorn` dans le `Procfile`.

> Certaines plateformes (comme Render) n'exigent pas de `Procfile` : elles demandent directement une « Start Command » dans leur interface, où tu écris `gunicorn app:app`.

Ensuite, le **déploiement continu** fonctionne tout seul : à chaque `git push` sur GitHub, Render redéploie automatiquement la nouvelle version. 🎉

```
git push  ->  GitHub  ->  (webhook)  ->  Render redéploie  ->  site à jour
```

````python
%%writefile app.py
import os
from flask import Flask, jsonify
from flask_sqlalchemy import SQLAlchemy
from dotenv import load_dotenv

load_dotenv()

app = Flask(__name__)
app.secret_key = os.environ.get("SECRET_KEY", "secours-dev")
app.config["SQLALCHEMY_DATABASE_URI"] = os.environ.get("DATABASE_URL", "sqlite:///blog.db")
db = SQLAlchemy(app)

class Article(db.Model):
    id = db.Column(db.Integer, primary_key=True)
    titre = db.Column(db.String(200), nullable=False)

@app.route("/")
def accueil():
    return jsonify({"message": "Blog prêt pour la production",
                    "nb_articles": Article.query.count()})

if __name__ == "__main__":
    with app.app_context():
        db.create_all()
    port = int(os.environ.get("PORT", 5000))
    debug = os.environ.get("DEBUG", "False") == "True"
    app.run(host="0.0.0.0", port=port, debug=debug)
````
